PyTorch의 torch.nn 모듈을 이용하여 신경망을 구축하기

10차원의 입력을 받아서 50차원의 은닉층을 거쳐 1차원의 출력을 생성하는 간단한 신경망을 정의

```python
import torch.nn as nn
import torch.nn.functional as F

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(10, 50)
        self.fc2 = nn.Linear(50, 1)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# 모델 인스턴스 생성
model = SimpleNN()
```

데이터셋을 처리하고 배치, 셔플링하기

가상의 데이터셋을 생성하고 데이터를 배치 단위로 나누어 처리하는 과정

```python
from torch.utils.data import DataLoader, TensorDataset

# 샘플 데이터 생성
x_data = torch.randn(100, 10)
y_data = torch.randn(100, 1)

# TensorDataset 및 DataLoader
dataset = TensorDataset(x_data, y_data)
dataloader = DataLoader(dataset, batch_size=5, shuffle=True)

# DataLoader 사용
for batch_idx, (x_batch, y_batch) in enumerate(dataloader):
    print(f'Batch {batch_idx + 1}:')
    print(x_batch)
    print(y_batch)
```

손실 함수 및 최적화 기법 사용

평균제곱오차(MSE)를 손실함수로 사용하고, 확률적 경사하강법(SGD)를 최적화 방법으로 이용하여 모델을 학습시키는 과정

```python
import torch.optim as optim

# 손실 함수 및 옵티마이저 설정
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 학습 루프
for epoch in range(10):
    for x_batch, y_batch in dataloader:
        # 순전파
        outputs = model(x_batch)
        loss = criterion(outputs, y_batch)

        # 역전파 및 최적화
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f'Epoch {epoch+1}, Loss: {loss.item()}')
```

모델의 성능을 평가하고 학습된 모델을 저장 및 불러오기

저장된 모델을 불러와 새로운 데이터에 대한 예측을 할 수 있음.

```python
# 평가 모드로 전환
model.eval()

# 평가
with torch.no_grad():
    total_loss = 0
    for x_batch, y_batch in dataloader:
        outputs = model(x_batch)
        loss = criterion(outputs, y_batch)
        total_loss += loss.item()

print(f'Total Loss: {total_loss}')

# 모델 저장
torch.save(model.state_dict(), 'simple_nn.pth')
```

```python
# 모델 로드
model = SimpleNN()
model.load_state_dict(torch.load('simple_nn.pth'))
model.eval()

# 예측 수행
with torch.no_grad():
    new_data = torch.randn(1, 10)
    prediction = model(new_data)
    print(f'Prediction: {prediction}')
```

```python
import torch

# 1. 차원 축소 (Squeeze)
tensor = torch.randn(1, 3, 1, 4)
print("Original Tensor:", tensor.shape)

tensor_squeezed = tensor.squeeze()
print("Squeezed Tensor:", tensor_squeezed.shape)

# 2. 차원 확장 (Unsqueeze)
tensor_expanded = tensor_squeezed.unsqueeze(0)
print("Expanded Tensor:", tensor_expanded.shape)

# 3. 축 변경 (Permute)
tensor_permuted = tensor.permute(1, 0, 2, 3)
print("Permuted Tensor:", tensor_permuted.shape)

# 4. 행렬 붙이기 (Concatenate)
tensor_a = torch.randn(2, 3)
tensor_b = torch.randn(2, 3)
tensor_concat = torch.cat((tensor_a, tensor_b), dim=0)
print("Concatenated Tensor:", tensor_concat.shape)

# 5. 행렬 간 연산
tensor_c = torch.randn(2, 3)
tensor_d = torch.randn(2, 3)

# 덧셈
tensor_add = tensor_c + tensor_d
print("Added Tensor:\n", tensor_add)

# 곱셈 (요소별 곱)
tensor_mul = tensor_c * tensor_d
print("Multiplied Tensor:\n", tensor_mul)

# 행렬 곱셈
tensor_matmul = torch.matmul(tensor_c, tensor_d.T)
print("Matrix Multiplication:\n", tensor_matmul)
```
